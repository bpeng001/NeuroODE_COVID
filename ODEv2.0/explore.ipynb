```python
# =========================
# Vaccine- and immunity-related parameters
# =========================

# Reduction in infection susceptibility after 2024-2025 COVID vaccination.
# Conservative range for late-2024 to early-2025 circulation.
# https://www.cdc.gov/mmwr/volumes/74/wr/mm7406a1.htm
# https://www.nature.com/articles/s41467-025-67796-0
epsilon1_lb = 0.20
epsilon1_ub = 0.35
epsilon1_pe = (epsilon1_lb + epsilon1_ub) / 2.0

# Additional reduction in hospitalization progression for immune / breakthrough infections.
# Kept below overall hospitalization VE because infection protection is already modeled by epsilon1.
# https://www.cdc.gov/mmwr/volumes/74/wr/mm7406a1.htm
# https://www.nature.com/articles/s41467-025-67796-0
epsilon2 = 0.25

# Loss of strong post-infection protection / return toward reinfection risk.
# CDC notes protection from reinfection lasts for several months; 120 days is a conservative default.
# https://www.cdc.gov/covid/about/reinfection.html
reinf = 1 / 120

# Fraction moving from recovered to partially susceptible rather than protected.
# Scenario / calibration parameter; no direct external estimate is used here.
re_pro = 0.20

# Duration of stronger post-infection protection before waning.
# In the current simulator, wan acts on S_A_prot / S_B_prot rather than V.
# https://www.cdc.gov/covid/about/reinfection.html
# https://www.nature.com/articles/s41467-025-67796-0
waning_time_lb = 90
waning_time_ub = 180
waning_time_pe = (waning_time_lb + waning_time_ub) / 2
wan = 1.0 / waning_time_pe

# Placeholder for a more detailed waned-protection structure.
# Not used in the current simulator.
ve_reduction_lb = 0.40
ve_reduction_ub = 0.60
ve_reduction_pe = (ve_reduction_lb + ve_reduction_ub) / 2

# Cross-immune escape assumptions.
# During 2024-2025, vaccination is treated as KP.3.1.1-specific, and all non-KP variants share the same
# vaccine escape penalty. During 2025-2026, vaccination is treated as XFG-specific.
# LP.8.1 was assessed by WHO as having immune escape comparable to XEC, so use a moderate scenario value
# for the fitting epochs. The post-fit XFG emergence is assigned a larger escape value as a scenario to
# allow a faster and higher rebound wave.
# https://www.who.int/publications/m/item/risk-evaluation-for-sars-cov-2-variant-under-monitoring-lp81
immune_escape_lb = 0.10
immune_escape_ub = 0.30
immune_escape_pe = (immune_escape_lb + immune_escape_ub) / 2
immune_escape_xfg = 0.45

# =========================
# Other model parameters
# =========================

# Total population
N = 31_290_831

# Latent rate (E -> I), sigma
sigma_lb = 1.0 / 3.0
sigma_ub = 1.0 / 1.0
sigma_pe = (sigma_lb + sigma_ub) / 2.0

# Recovery rate (I -> R), gamma
gamma_lb = 1.0 / 7
gamma_ub = 1.0 / 5
gamma_pe = (gamma_lb + gamma_ub) / 2.0

# Recovery rate (H -> R)
gam2 = 1 / (7 + 14)

# Infectious to Hospitalization rate (IHR)
IHR_lb = 0.003
IHR_ub = 0.01
IHR_pe = (IHR_lb + IHR_ub) / 2.0

delta = 1 / (8 + 10) #duration (+latent)

# Hospitalization to Death rate (HDR)
HDR = 0.16

eta = 1 / (10 + 14) #duration

# Reporting rate
rep = 0.2

# Approximate 2024-2025 COVID vaccine coverage level; not currently used in the simulator.
# https://www.nature.com/articles/s41467-025-67796-0
cov = 0.23

transmission_epochs = [
    {
        "start_date": "2024-11-09",
        "end_date": "2024-12-20",
        "a_share_column": "KP.3.1.1_share",
        "b_share_column": "XEC_share",
        "beta_multiplier_b": 1.0,
        "immune_escape_history_b_on_a": immune_escape_pe,
        "immune_escape_vaccine_mismatch": immune_escape_pe,
    },
    {
        "start_date": "2024-12-21",
        "end_date": "2025-04-26",
        "a_share_column": "XEC_share",
        "b_share_column": "LP.8.1_share",
        "beta_multiplier_b": 1.0,
        "immune_escape_history_b_on_a": immune_escape_pe,
        "immune_escape_vaccine_mismatch": immune_escape_pe,
    },
    {
        "start_date": "2025-04-27",
        "end_date": None,
        "a_share_column": "LP.8.1_share",
        "b_share_column": "XFG_share",
        "beta_multiplier_b": 1.25,
        "immune_escape_history_b_on_a": immune_escape_xfg,
        "immune_escape_vaccine_mismatch": immune_escape_xfg,
    },
]

vaccine_target_epochs = [
    {
        "start_date": "2024-10-01",
        "end_date": "2025-09-30",
        "target_share_column": "KP.3.1.1_share",
    },
    {
        "start_date": "2025-10-01",
        "end_date": None,
        "target_share_column": "XFG_share",
    },
]

variant_cfg = {
    "share_start_date": pd.Timestamp(transmission_epochs[0]["start_date"]),
    "transmission_epochs": transmission_epochs,
    "vaccine_target_epochs": vaccine_target_epochs,
    "immune_escape_history_B_on_A": immune_escape_pe,
    "immune_escape_vaccine_mismatch": immune_escape_pe,
    "waning_multiplier_B": 2.0,
    "beta_multiplier_B": 1.0,
}
```

```python
fit_horizon_days = 180
projection_horizon_days = 410 + 510
projection_week_index = range(129)
fit_start_date = pd.Timestamp(GC_['date'].iloc[0])
projection_start_date = pd.Timestamp(GC_1['date'].iloc[0]) if 'GC_1' in globals() else None


def make_model_params(wan_override=None, reinf_override=None):
    return {
        "N": N,
        "epsilon1": epsilon1_pe,
        "epsilon2": epsilon2,
        "IHR": IHR_pe,
        "HDR": HDR,
        "gamma1": gamma_pe,
        "gamma2": gam2,
        "delta": delta,
        "eta": eta,
        "rep": rep,
        "re_pro": re_pro,
        "reinf": reinf if reinf_override is None else reinf_override,
        "wan": wan if wan_override is None else wan_override,
        "V0": float(vacc_ts["total_vaccinated"].loc[0]),
        "Inh0": 160_000 * IHR_pe,
        "Inr0": 160_000 * (1 - IHR_pe),
        "Imh0": 80_000 * IHR_pe,
        "Imr0": 80_000 * (1 - IHR_pe),
        "R0": 200_000,
        "Hd0": 10_000 * HDR,
        "Hr0": 10_000 * (1 - HDR),
        "D0": 70,
        "SxS0": 6_000_000 * (1 - re_pro),
        "SxI0": 6_000_000 * re_pro,
    }


def make_variant_share_series(start_date, horizon_days):
    daily_index = pd.date_range(pd.Timestamp(start_date), periods=horizon_days, freq="D")
    return build_variant_share_series(
        variant_marker,
        daily_index,
        variant_cfg["share_start_date"],
        variant_cfg["transmission_epochs"][0]["b_share_column"],
        epoch_schedule=variant_cfg["transmission_epochs"],
    )


def run_variant_sim(
    beta,
    horizon_days,
    weekly_index,
    start_date,
    wan_override=None,
    reinf_override=None,
    stochastic_cfg=None,
    return_states=False,
):
    model_params = make_model_params(wan_override=wan_override, reinf_override=reinf_override)
    variant_share = make_variant_share_series(start_date, horizon_days)
    return simulate_variant_model(
        beta=beta,
        horizon_days=horizon_days,
        weekly_index=weekly_index,
        vacc_series=y_vac,
        variant_share_series=variant_share,
        model_params=model_params,
        variant_cfg=variant_cfg,
        stochastic_cfg=stochastic_cfg,
        return_states=return_states,
    )


def sim_fit(x, beta):
    result = run_variant_sim(beta, fit_horizon_days, time, fit_start_date)
    return result["weekly_hosp"][:-1]


def sim_det(x, beta):
    result = run_variant_sim(beta, fit_horizon_days, time, fit_start_date)
    return result["weekly_hosp"]


def sim_fit1(x, beta, wan1):
    result = run_variant_sim(beta, fit_horizon_days, time, fit_start_date, wan_override=wan1)
    return result["weekly_hosp"][:-1]


def sim_det1(x, beta, wan1):
    result = run_variant_sim(
        beta,
        projection_horizon_days,
        projection_week_index,
        projection_start_date,
        wan_override=wan1,
    )
    return result["weekly_hosp"]


def sim_fit2(x, beta, reinf1):
    result = run_variant_sim(beta, fit_horizon_days, time, fit_start_date, reinf_override=reinf1)
    return result["weekly_hosp"][:-1]


def sim_det2(x, beta, reinf1):
    result = run_variant_sim(
        beta,
        projection_horizon_days,
        projection_week_index,
        projection_start_date,
        reinf_override=reinf1,
    )
    return result["weekly_hosp"]


def sim_sto(x, beta_bar, std, kappa, seed=None):
    result = run_variant_sim(
        beta_bar,
        projection_horizon_days,
        projection_week_index,
        projection_start_date,
        stochastic_cfg={"std": std, "kappa": kappa, "seed": seed},
    )
    return result["weekly_hosp"]
```
